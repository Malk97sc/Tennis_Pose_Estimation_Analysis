```python
from roboflow import Roboflow
from ultralytics import YOLO
```

```python
api = " "

rf = Roboflow(api_key = api)
project = rf.workspace("me-dxtf3").project("tennis-ball-tracking-detection-h9rat")
version = project.version(1)
dataset = version.download("yolov8")
```

```python
model = YOLO("yolov8x.pt")

data = "Tennis-Ball-Tracking-Detection--1/data.yaml"

epochs = 10
image_size = 640
batch = 64
name_folder = "yolo"
name_model = "ball_yolov8x"

model.train(data = data,
            epochs = epochs,
            imgsz = image_size,
            batch = batch,
            project = name_folder,
            name = name_model,
            device = 0
            )
```
